Small-world graphs are characterized by the following three properties:
- They are _sparse_: the number of vertices is much smaller than the number of edges.
- The have _short average path lengths_: if you pick two random vertices, the length of the shortest path between them is short.
- They exhibit _local clustering_: if two vertices are neighbors of a third vertex, then the two vertices are likely to be neighbors of each other.


The term _small world_ refers to the idea that the preponderance of vertices have both local clustering and short paths to other vertices.   
The modifier _phenomeon_ refers to the unexpected fact that so many graphs that arise in practice are sparse, have short paths, and exhibit local clustering.
