```python
c1 = Charge(.51, .63, 21.3)
c2 = Charge(.51, .63, 21.3)
c3 = c1

print(f"c1 == c2 ?", c1 == c2)
print(f"c1 == c3 ?", c1 == c3)
print(f"id(c1) =", id(c1))
print(f"id(c2) =", id(c2))
print(f"id(c2) =", id(c3))
print(f"c1 is c2 ?", c1 is c2)
print(f"c1 is c3 ?", c1 is c3)
```

```python
a = 123_456_789
b = 123_456_789
c = a
print(f"a == b ?", a == b)
print(f"a == c ?", a == c)
print(f"id(a) =", id(a))
print(f"id(b) =", id(b))
print(f"id(c) =", id(c))
print(f"a is b ?", a is b)
print(f"a is c ?", a is c)
```
