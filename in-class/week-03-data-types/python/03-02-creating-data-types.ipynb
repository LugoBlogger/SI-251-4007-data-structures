### Complex numbers

The basic properties of complex numbers
1. Addition: $(x + y\mathbb{i}) + (v + wi) = (x + v) + (y + w)i $
2. Multiplication: $(x + yi) * (v + wi) = (xv - yw)+ (yv + xw) i$ 
3. Magnitude: $|x + yi| = \sqrt{x^2 + y^2}$
4. Real part: $\operatorname{\mathrm{Re}}(x + yi) = x$
5. Imaginarty part: $\operatorname{\mathrm{Im}}(x + yi) = y$
