The playwright Samuel Beckett, perhaps best known for _Waiting for Godot_,
wrote a play called _Quad_ that had the following property: starting with an
empty stage, characters enter and exit at a time so that each subset of characters 
on the stage appears exactly once. How did Becket generate the stage directions
for this play?

An $n$-bit _Gray code_ is a list of the $2^n$ diffeerent $n$-bit binary numbers 
such that each entry in the list differs in precisely one bit from its predecessor.

How do we generate a Gray code?   
The $n$-bit binary _reflected Gray code_ is defined recursively as follows:
* The $(n-1)$ bit code, with 0 prepended to each codeword, followed by
* The $(n-1)$ bit code _in reverse order_, with 1 prepended to each codeword
