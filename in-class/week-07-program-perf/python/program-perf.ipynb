```python
start_t = time.perf_counter()

n = 200_000
a = []
for i in range(n):
  a.insert(0, "slow")

end_t = time.perf_counter()
print(f"running time: {end_t - start_t}")
```

```python
start_t = time.perf_counter()

n = 200_000
a = []
for i in range(n):
  a.insert(i, "fast")

end_t = time.perf_counter()
print(f"running time: {end_t - start_t}")
```

* Integers

```python
print(f"2^30 - 1 = {2**30 - 1:,}")
print(f"2^60 - 1 = {2**60 - 1:,}")
```

```python
print(f"                                           size in bytes")
print(f"sys.getsizeof(0)                           ", size(0))

print(f"sys.getsizeof(123,456,789)                 ", size(123_456_789))

print(f"sys.getsizeof(-1,073,741,824) = -2^30      ", size(-1_073_741_824))
print(f"sys.getsizeof(-1,073,741,823) = -(2^30 - 1)", size(-1_073_741_823))
print(f"sys.getsizeof(1,073,741,823) = (2^30 - 1)  ", size(1_073_741_823))
print(f"sys.getsizeof(1,073,741,824) = (2^30)      ", size(1_073_741_824))
print(f"sys.getsizeof(123,456,789,123)             ", size(123_456_789_123))

print(f"sys.getsizeof(1,152,921,504,606,846,975)   ", size(1_152_921_504_606_846_975))
print(f"sys.getsizeof(1,152,921,504,606,846,976)   ", size(1_152_921_504_606_846_976))
print(f"sys.getsizeof(123456789_123456789)         ", size(1234567890_1234567890))
```
