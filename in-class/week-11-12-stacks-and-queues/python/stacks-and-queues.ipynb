### Python list (resizing array) implementation of a stack

Test if it is faster to insert new item at the end of a stack (a list)
than at the start of a stack

```python
rng = np.random.default_rng()
arr = rng.integers(1_000_000, size=200_000)
```

```python
# push an item to the stack at the end of a stack (list)
start_t = time.perf_counter()

b = []
for val in arr:
  b.insert(-1, val)

comp_t = time.perf_counter() - start_t
comp_t
```

```python
# push an item to the stack at the beginning of a stack (list)
# comp_time: 
start_t = time.perf_counter()

b = []
for val in arr:
  b.insert(0, val)

comp_t = time.perf_counter() - start_t
comp_t
```
