### Python list (resizing array) implementation of a stack

Test if it is faster to insert new item at the end of a stack (a list)
than at the start of a stack

```python
rng = np.random.default_rng()
arr = rng.integers(1_000_000, size=200_000)
```

```python
# push an item to the stack at the end of a stack (list)
start_t = time.perf_counter()

b = []
for val in arr:
  b.insert(-1, val)

comp_t = time.perf_counter() - start_t
comp_t
```

```python
# push an item to the stack at the beginning of a stack (list)
# comp_time: 
start_t = time.perf_counter()

b = []
for val in arr:
  b.insert(0, val)

comp_t = time.perf_counter() - start_t
comp_t
```

### Queue applications

* M/M/1 queue

To understand M/M/1 queue, we have to know about exponential distribution.
Check the notebook [`exponential-distribution.ipynb`](./exponential-distribution.ipynb).

The reason why we set M in M/M/1 which is stand for Markovian (using exponential distribution) because this distribution has memoryless property.


M/M/1 queue is parameterized by its _arrival rate_ $\lambda$ (for example, 
the number of cars per minute arriving at the toll booth), and its _service rate_
$\mu$ (for example, the number of cars per minute that can pass through the toll
booth)

M/M/1 is characterized by three properties:
- There is one server&mdash;a FIFO queue
- Interarrival times to a queue obey an exponential distribution with rate $\lambda$ per minute
- Service times from a nonempty queue obey an exponential distribution with rate
  $\mu$ per minute

The average time between arrivals is $1/\lambda$ minutes and the average time between services (when the queue is nonempty) is $1/\mu$ per minutes.
So, the queue will grow without bound (to infinity) if $1/\lambda < 1/\mu$ or $\lambda > \mu$.
The interesting dynamic process will happen if we have $\lambda < \mu$

* Analyze

In the probability theory, it can be shown that for M/M/1 queues, we have    
a. The average number of customers in the system, $L$ is $\lambda / (\mu - \lambda)$    
b. The average time a customer spends in the system, $W$ is $1 / (\mu - \lambda)$
